```python
from pathlib import Path

import numpy as np
import soundfile as sf
```

```python
CLEAN_SIGNAL_PATH = Path("../integration/fixtures/clean_speech_1.wav")
clean_signal, sr = sf.read(CLEAN_SIGNAL_PATH)
```

```python
def add_white_noise(
    clean: np.ndarray, seed: int, snr_db: float = 0.0
) -> tuple[np.ndarray, np.ndarray]:
    rng = np.random.default_rng(seed)
    raw_noise = rng.standard_normal(clean.shape)

    signal_power = np.mean(clean**2)
    noise_power = np.mean(raw_noise**2)

    target_noise_power = signal_power / (10 ** (snr_db / 10))
    raw_noise *= np.sqrt(target_noise_power / noise_power)
    noisy_signal = clean + raw_noise
    return noisy_signal, raw_noise
```

```python
input_signal, noise_ref = add_white_noise(clean_signal, 42)
```

# Using the Filter

```python
from adaptif import LMSFilter
```

```python
lms = LMSFilter(step_size=0.1, window_size=16)

output_signal = lms.adapt(input_signal, noise_ref)
```

```python
from IPython.display import Audio, display

print("Input Signal:")
display(Audio(input_signal, rate=sr))

print("Noise Reference:")
display(Audio(noise_ref, rate=sr))

print("Output Signal:")
display(Audio(output_signal, rate=sr))
```
